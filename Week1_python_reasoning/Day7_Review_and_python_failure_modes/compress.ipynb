### Runtime Error

Bad

```python
def is_adult(age):
    return age >= 18
```

Better

```python
def is_adult(age):
    if age is None:
        raise ValueError("age cannot be None")

    if not isinstance(age, (int, float)):
        raise TypeError("age must be a number")

    return age >= 18
```

## Hidden state

```python
data = []

def add_item(item):
    data.append(item)
```

Better:

```python
def add_item(items, item):
    return items + [item]
```

#### Another hidden step trap:

```python
def add_item(item, items=[]):
    items.append(item)
    return items

print(add_item(5))  # [5]
print(add_item(8))  # [5,8]
print(add_item(10)) # [5,8,10] <--- same list reused
```

Better:

```python
def add_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

print(add_item(1))  # [1]
print(add_item(2))  # [2]  <-- new list created
```

## 5. Weak Error Handling

Bad:

```python
def get_first_item(items):
    return items[0]
```

Better:

```python
def get_first_item(items):
    if items is None:
        raise ValueError("items cannot be None")

    if len(items) == 0:
        raise ValueError("items cannot be empty")

    return items[0]
```
