## Clean version

```python
import pandas as pd


DATA_PATH = "students.csv"
REQUIRED_COLUMNS = ["name", "age", "score"]
PASS_MARK = 50


def load_data(path):
    """Load a CSV file and return a DataFrame."""
    return pd.read_csv(path)


def validate_columns(df, required_columns):
    """Check that all required columns exist in the DataFrame."""
    missing_columns = []

    for column in required_columns:
        if column not in df.columns:
            missing_columns.append(column)

    if missing_columns:
        raise ValueError(f"Missing required columns: {missing_columns}")


def clean_data(df):
    """Remove rows with missing values."""
    return df.dropna()


def calculate_average_age(df):
    """Calculate the average age."""
    return df["age"].mean()


def calculate_average_score(df):
    """Calculate the average score."""
    return df["score"].mean()


def get_passing_students(df, pass_mark):
    """Return students whose score is greater than or equal to the pass mark."""
    return df[df["score"] >= pass_mark]


def summarize_data(df):
    """Return summary statistics for numeric columns."""
    return df.describe()


def main():
    raw_data = load_data(DATA_PATH)

    validate_columns(raw_data, REQUIRED_COLUMNS)

    clean_students = clean_data(raw_data)

    average_age = calculate_average_age(clean_students)
    average_score = calculate_average_score(clean_students)
    passing_students = get_passing_students(clean_students, PASS_MARK)
    summary = summarize_data(clean_students)

    print("Summary:")
    print(summary)

    print("\nAverage age:")
    print(average_age)

    print("\nAverage score:")
    print(average_score)

    print("\nPassing students:")
    print(passing_students)


# if __name__ == "__main__":
#     main()
```
