## Building the Day 6 project

```python
import pandas as pd

REQUIRED_COLUMNS = ["name", "age", "score"]

def load_csv(path):
    """Load a CSV file into a pandas DataFrame."""
    return pd.read_csv(path)


def validate_columns(df, required_columns):
    """Check whether required columns exist in the DataFrame."""
    missing_columns = []

    for column in required_columns:
        if column not in df.columns:
            missing_columns.append(column)

    if missing_columns:
        raise ValueError(f"Missing required columns: {missing_columns}")


def summarize_dataframe(df):
    """Print basic information about the DataFrame."""
    print("Rows and columns:")
    print(df.shape)

    print("\nColumn names:")
    print(list(df.columns))

    print("\nMissing values:")
    print(df.isna().sum())

    print("\nNumeric summary:")
    print(df.describe())


def main():
    file_path = "data/sample.csv"

    df = load_csv(file_path)
    validate_columns(df, REQUIRED_COLUMNS)
    summarize_dataframe(df)


# if __name__ == "__main__":
#     main()
```

### Improved Day 6 version

```python
import pandas as pd


REQUIRED_COLUMNS = ["name", "age", "score"]
NUMERIC_COLUMNS = ["age", "score"]
DATA_PATH = "../../data/student.csv"


def load_csv(path):
    """Load a CSV file into a pandas DataFrame."""
    return pd.read_csv(path)


def normalize_column_names(df):
    """Remove extra spaces and lowercase column names."""
    df = df.copy()
    df.columns = df.columns.str.strip().str.lower()
    return df


def validate_columns(df, required_columns):
    """Raise an error if required columns are missing."""
    missing_columns = []

    for column in required_columns:
        if column not in df.columns:
            missing_columns.append(column)

    if missing_columns:
        raise ValueError(f"Missing required columns: {missing_columns}")


def validate_numeric_columns(df, numeric_columns):
    """Check whether expected numeric columns contain numeric values."""
    for column in numeric_columns:
        converted_column = pd.to_numeric(df[column], errors="coerce")
        invalid_count = converted_column.isna().sum() - df[column].isna().sum()

        if invalid_count > 0:
            raise ValueError(f"Column '{column}' contains {invalid_count} invalid numeric values")


def summarize_dataframe(df):
    """Print basic information about the DataFrame."""
    print("Rows and columns:")
    print(df.shape)

    print("\nColumn names:")
    print(list(df.columns))

    print("\nData types:")
    print(df.dtypes)

    print("\nMissing values:")
    print(df.isna().sum())

    print("\nNumeric summary:")
    print(df.describe())


def main():
    df = load_csv(DATA_PATH)
    df = normalize_column_names(df)

    validate_columns(df, REQUIRED_COLUMNS)
    validate_numeric_columns(df, NUMERIC_COLUMNS)

    summarize_dataframe(df)


# if __name__ == "__main__":
#     main()
```
